# 📊 Análise de Vendas com Python

## Sobre o projeto

Este projeto apresenta uma análise exploratória de dados fictícios de vendas de produtos de informática, utilizando Python.

O objetivo é identificar o faturamento por produto, a quantidade de unidades vendidas e os produtos com melhor desempenho comercial.

## Tecnologias utilizadas

- Python
- Pandas — manipulação e análise de dados
- Matplotlib — visualização de dados
- Ipywidgets — filtros interativos
- Google Colab — ambiente de desenvolvimento

## Principais resultados

- **Faturamento total:** R$ 22.800,00
- **Unidades vendidas:** 43
- **Produto com maior faturamento:** Notebook
- **Produto com maior quantidade vendida:** Mouse

## Funcionalidades

O dashboard permite selecionar produtos, visualizar indicadores de vendas e comparar o faturamento por meio de gráficos.

```python
import pandas as pd

dados = {
    "Produto": ["Notebook", "Mouse", "Teclado", "Monitor"],
    "Quantidade": [5, 20, 10, 8],
    "Preco": [2500, 80, 150, 900]
}

tabela = pd.DataFrame(dados)

print(tabela)
```

```python
tabela["Faturamento"] = tabela["Quantidade"] * tabela["Preco"]

display(tabela)
```

```python
import matplotlib.pyplot as plt

plt.bar(tabela["Produto"], tabela["Faturamento"])

plt.title("Faturamento por Produto")
plt.xlabel("Produtos")
plt.ylabel("Faturamento em R$")

plt.show()
```

```python
faturamento_total = tabela["Faturamento"].sum()

print("Faturamento total: R$", faturamento_total)
```

```python
total_produtos = tabela["Quantidade"].sum()

print("Total de unidades vendidas:", total_produtos)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(tabela["Produto"], tabela["Faturamento"])

ax.set_title("Dashboard de Vendas")
ax.set_xlabel("Produtos")
ax.set_ylabel("Faturamento (R$)")

fig.text(0.15, 0.95, f"Faturamento: R$ {faturamento_total:,.2f}")
fig.text(0.60, 0.95, f"Unidades vendidas: {total_produtos}")

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 6))

# Espaço para os indicadores
fig.subplots_adjust(top=0.72)

# Título principal
fig.suptitle("Dashboard de Vendas", fontsize=18, fontweight="bold")

# Indicadores
fig.text(0.15, 0.82, f"Faturamento Total\nR$ {faturamento_total:,.2f}",
         fontsize=13, fontweight="bold")

fig.text(0.65, 0.82, f"Unidades Vendidas\n{total_produtos}",
         fontsize=13, fontweight="bold")

# Gráfico
ax.bar(tabela["Produto"], tabela["Faturamento"])

ax.set_title("Faturamento por Produto")
ax.set_ylabel("Faturamento (R$)")
ax.set_xlabel("Produtos")

plt.show()
```

```python
produto_mais_vendido = tabela.loc[tabela["Quantidade"].idxmax(), "Produto"]

print("Produto mais vendido:", produto_mais_vendido)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 6))

fig.subplots_adjust(top=0.70)

# Título do dashboard
fig.suptitle("Dashboard de Vendas", fontsize=20, fontweight="bold")

# Indicadores
fig.text(0.12, 0.82, "Faturamento Total", fontsize=12)
fig.text(0.12, 0.77, f"R$ {faturamento_total:,.2f}", fontsize=15, fontweight="bold")

fig.text(0.43, 0.82, "Unidades Vendidas", fontsize=12)
fig.text(0.43, 0.77, str(total_produtos), fontsize=15, fontweight="bold")

fig.text(0.73, 0.82, "Mais Vendido", fontsize=12)
fig.text(0.73, 0.77, produto_mais_vendido, fontsize=15, fontweight="bold")

# Gráfico
ax.bar(tabela["Produto"], tabela["Faturamento"])

ax.set_title("Faturamento por Produto")
ax.set_ylabel("Faturamento (R$)")
ax.set_xlabel("Produtos")

plt.show()
```

```python
# Organizar os produtos pelo faturamento
ranking = tabela.sort_values(
    by="Faturamento",
    ascending=False
)

# Mostrar o ranking
display(ranking)
```

```python
import matplotlib.pyplot as plt

# Criar gráfico horizontal
plt.figure(figsize=(10, 5))

plt.barh(
    ranking["Produto"],
    ranking["Faturamento"],
    color="steelblue"
)

# Colocar o maior faturamento no topo
plt.gca().invert_yaxis()

# Títulos
plt.title("Ranking de Faturamento por Produto")
plt.xlabel("Faturamento (R$)")
plt.ylabel("Produtos")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, HTML

# Dados das vendas
tabela = pd.DataFrame({
    "Produto": ["Notebook", "Mouse", "Teclado", "Monitor"],
    "Quantidade": [5, 20, 10, 8],
    "Preco": [2500, 80, 150, 900]
})

tabela["Faturamento"] = tabela["Quantidade"] * tabela["Preco"]

# Criar filtro
filtro = widgets.Dropdown(
    options=["Todos"] + tabela["Produto"].tolist(),
    description="Produto:"
)

# Criar dashboard
def painel(produto):

    if produto == "Todos":
        dados = tabela
    else:
        dados = tabela[tabela["Produto"] == produto]

    faturamento = dados["Faturamento"].sum()
    unidades = dados["Quantidade"].sum()

    display(HTML(f"""
    <h2>📊 Dashboard de Vendas</h2>
    <div style="display:flex; gap:15px;">
        <div style="background:#2563eb; padding:20px; border-radius:12px;">
            <h3>Faturamento</h3>
            <h2>R$ {faturamento:,.2f}</h2>
        </div>
        <div style="background:#059669; padding:20px; border-radius:12px;">
            <h3>Unidades Vendidas</h3>
            <h2>{unidades}</h2>
        </div>
    </div>
    """))

    plt.figure(figsize=(9, 5))

    barras = plt.bar(
        dados["Produto"],
        dados["Faturamento"],
        color="#3b82f6"
    )

    plt.bar_label(barras, padding=3)
    plt.title("Faturamento por Produto")
    plt.ylabel("Faturamento (R$)")
    plt.ylim(0, 14000)
    plt.show()

# Ativar dashboard interativo
widgets.interactive(
    painel,
    produto=filtro
)
```
